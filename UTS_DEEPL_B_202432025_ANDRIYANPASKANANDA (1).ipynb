<b>Konsep:</b> LSTM mempunyai sel memori (C) dan hidden state (h). Forget gate memilih informasi lama yang dipertahankan, input gate memilih informasi baru yang masuk, dan output gate menentukan hidden state. Dalam pemodelan teks, setiap kata dipetakan menjadi vektor embedding sebelum diproses secara berurutan.
<br><br><b>Contoh manual ilustratif: satu nilai input dan satu nilai output pada t=1</b><br>Dipilih nilai sederhana x<sub>t</sub> = 0,6; h<sub>t−1</sub> = 0,2; C<sub>t−1</sub> = 0,1. Semua bobot dan bias di bawah adalah <b>contoh untuk belajar, bukan bobot model hasil training</b>.<br>
1. f<sub>t</sub> = σ(0,4(0,6)+0,3(0,2)−0,1) = σ(0,20) = <b>0.5498</b><br>
2. i<sub>t</sub> = σ(0,5(0,6)−0,2(0,2)) = σ(0,26) = <b>0.5646</b><br>
3. C̃<sub>t</sub> = tanh(0,6(0,6)+0,1(0,2)+0,05) = tanh(0,43) = <b>0.4053</b><br>
4. C<sub>t</sub> = f<sub>t</sub> × 0,1 + i<sub>t</sub> × C̃<sub>t</sub> = <b>0.2838</b><br>
5. o<sub>t</sub> = σ(0,3(0,6)+0,4(0,2)+0,05) = σ(0,31) = <b>0.5769</b><br>
6. h<sub>t</sub> = o<sub>t</sub> × tanh(C<sub>t</sub>) = <b>0.1595</b> (satu hidden-state output).<br><br>
<b>Catatan:</b> σ(z)=1/(1+e<sup>−z</sup>); tanh(z)=(e<sup>z</sup>−e<sup>−z</sup>)/(e<sup>z</sup>+e<sup>−z</sup>). Implementasi yang sebenarnya memakai vektor/tensor dan bobot yang dipelajari melalui backpropagation.
</div></div>

```python
#202432025_AndriyanPaskaNanda
import os  # Membaca konfigurasi lingkungan.
import json  # Menyimpan ringkasan metrik penelitian ke berkas JSON.
import random  # Mengatur pengacakan agar dapat direplikasi.
import re  # Melakukan tokenisasi dengan pola reguler.
from collections import Counter  # Menghitung frekuensi kata.
from pathlib import Path  # Mengatur lokasi berkas.
import numpy as np  # Operasi numerik dan array.
import pandas as pd  # Membaca dan mengolah dataset.
import matplotlib.pyplot as plt  # Membuat grafik.
import torch  # Framework deep learning.
import torch.nn as nn  # Mendefinisikan layer jaringan.
import torch.optim as optim  # Mengakses optimizer.
from torch.nn.utils.rnn import pack_padded_sequence  # Mengabaikan posisi padding di LSTM.
from torch.utils.data import TensorDataset, DataLoader  # Membentuk mini-batch.
from sklearn.model_selection import train_test_split  # Membagi data secara stratifikasi.
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score, accuracy_score  # Mengukur performa.
SEED = 42  # Menetapkan seed reproduksibilitas.
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)  # Menyetel seed.
if torch.cuda.is_available(): torch.cuda.manual_seed_all(SEED)  # Seed GPU bila tersedia.
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')  # Memilih GPU atau CPU.
print('Device:', DEVICE, '| PyTorch:', torch.__version__)  # Menampilkan lingkungan.
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>

```python
#202432025_AndriyanPaskaNanda
from zipfile import ZipFile  # Membaca ZIP dataset Kaggle tanpa perlu ekstraksi manual.

# Mencari CSV asli terlebih dahulu dalam sesi Colab saat ini.
file_paths = list(Path('.').glob('tripadvisor_hotel_reviews.csv')) + list(Path('.').glob('archive.zip'))  # Mendeteksi CSV atau ZIP yang telah diunggah pada sesi Colab.
if not file_paths:  # Meminta unggah sekali jika file belum ada.
    try:  # Fitur unggah dari Google Colab.
        from google.colab import files  # Menyediakan dialog pemilihan berkas.
        print('=== UNGGAH DATASET KAGGLE ===')  # Menandai langkah yang membutuhkan tindakan pengguna.
        print('Klik Choose Files, pilih ZIP unduhan Kaggle ATAU tripadvisor_hotel_reviews.csv.')  # Panduan unggah.
        uploaded = files.upload()  # Menampilkan jendela untuk memilih file dari komputer.
        file_paths = [Path(name) for name in uploaded if name.lower().endswith(('.csv', '.zip'))]  # Memilih CSV atau ZIP.
        if not file_paths:  # Mencegah training pada file yang salah.
            raise FileNotFoundError('Tidak ada berkas CSV/ZIP yang diunggah; silakan jalankan ulang cell.')  # Informasi kesalahan.
    except ImportError as exc:  # Alur untuk pengguna Jupyter lokal.
        raise FileNotFoundError('Di Jupyter lokal, taruh CSV/ZIP Kaggle di folder notebook.') from exc  # Pesan perbaikan.

# Memprioritaskan dataset dengan nama standar bila beberapa berkas tersedia.
source_path = next((p for p in file_paths if 'tripadvisor' in p.name.lower()), file_paths[0])  # Lokasi file terpilih.
if source_path.suffix.lower() == '.zip':  # Jika pengguna memilih ZIP unduhan Kaggle.
    with ZipFile(source_path) as archive:  # Membaca isi ZIP secara langsung.
        members = [name for name in archive.namelist() if name.lower().endswith('.csv') and not name.startswith('__MACOSX/')]  # Daftar CSV di ZIP.
        if not members:  # Menangani ZIP yang salah.
            raise ValueError('ZIP yang dipilih tidak berisi CSV. Unduh dataset Trip Advisor yang benar.')  # Petunjuk.
        member = next((m for m in members if 'tripadvisor' in m.lower()), members[0])  # Memilih CSV ulasan hotel.
        with archive.open(member) as csv_stream:  # Membuka file CSV di dalam ZIP.
            df = pd.read_csv(csv_stream, encoding='utf-8', encoding_errors='replace')  # Membaca data asli.
else:  # Jika pengguna memilih CSV yang sudah diekstrak.
    df = pd.read_csv(source_path, encoding='utf-8', encoding_errors='replace')  # Membaca CSV asli.

# Memvalidasi struktur dan menampilkan bukti eksplorasi dataset.
df.columns = df.columns.astype(str).str.strip()  # Membersihkan spasi pada nama kolom.
required_cols = {'Review', 'Rating'}  # Kolom wajib dataset Kaggle ini.
if not required_cols.issubset(df.columns):  # Memeriksa bahwa file yang dipilih benar.
    raise ValueError(f'CSV tidak cocok. Kolom ditemukan: {df.columns.tolist()}')  # Mencegah data tertukar.
print('Sumber data:', source_path.name)  # Menampilkan nama sumber untuk screenshot.
print('Data mentah (baris, kolom):', df.shape)  # Menampilkan ukuran dataset sebelum cleaning.
print('Kolom:', df.columns.tolist())  # Menampilkan atribut dataset.
print('Contoh data asli:')  # Pratinjau untuk laporan BAB 1.
print(df[['Review', 'Rating']].head(4).to_string(index=False, max_colwidth=75))  # Contoh teks dan rating.

# Menghapus data kosong, rating tidak valid, dan teks ulasan duplikat.
df = df[['Review', 'Rating']].copy()  # Memilih dua fitur yang digunakan.
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')  # Mengubah rating ke bilangan.
df = df.dropna(subset=['Review', 'Rating'])  # Menghapus baris tanpa teks/rating.
df = df[df['Rating'].isin([1, 2, 3, 4, 5])].copy()  # Membatasi skala bintang 1–5.
df['Review'] = df['Review'].astype(str).str.strip()  # Merapikan spasi ulasan.
df = df[df['Review'].str.len() > 0].drop_duplicates(subset='Review').reset_index(drop=True)  # Menghilangkan kosong dan duplikat.
df['sentiment'] = df['Rating'].map(lambda r: 0 if r <= 2 else (1 if r == 3 else 2))  # Label turunan rating.
LABELS = ['Negatif', 'Netral', 'Positif']  # Nama untuk tiga kelas.
print('Data setelah preprocessing:', len(df))  # Jumlah data bersih.
print('Distribusi sentimen:', df['sentiment'].value_counts().sort_index().rename(index=dict(enumerate(LABELS))))  # Frekuensi label.
assert df['sentiment'].nunique() == 3, 'Ketiga kategori harus ada. Periksa dataset Kaggle yang dipilih.'  # Validasi label.

# Memisahkan training/validation/test sebelum membangun vocabulary.
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['sentiment'], random_state=SEED)  # 70% train.
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['sentiment'], random_state=SEED)  # 15%+15%.
MAX_LEN, MAX_VOCAB, MIN_FREQ = 80, 12000, 2  # Pengaturan input sekuens dan kosakata.
def tokenize(text):  # Mengubah ulasan ke token bahasa Inggris.
    return re.findall(r"[a-z]+(?:'[a-z]+)?", text.lower())  # Lowercase dan tokenisasi regex.
counts = Counter(word for text in train_df['Review'] for word in tokenize(text))  # Hitung kata TRAIN saja.
vocab = {'<PAD>': 0, '<UNK>': 1}  # Identitas token untuk padding dan unknown.
for word, freq in counts.most_common():  # Mengurutkan kata menurut frekuensi.
    if len(vocab) >= MAX_VOCAB: break  # Menetapkan maksimum vocabulary.
    if freq >= MIN_FREQ: vocab[word] = len(vocab)  # Menambahkan token yang cukup sering.
def encode(text):  # Mengubah teks menjadi angka untuk model.
    ids = [vocab.get(word, 1) for word in tokenize(text)[:MAX_LEN]]  # Maksimal 80 token; OOV=1.
    size = max(1, len(ids))  # Minimum panjang satu untuk pack_padded_sequence.
    return (ids + [0] * (MAX_LEN - len(ids))), size  # Padding kanan hingga 80.
def make_loader(frame, shuffle=False):  # Menyiapkan batch data dengan panjang sekuens.
    encoded = [encode(text) for text in frame['Review']]  # Encoding semua ulasan.
    X = torch.tensor([x[0] for x in encoded], dtype=torch.long)  # Tensor indeks kata.
    lengths = torch.tensor([x[1] for x in encoded], dtype=torch.long)  # Tensor panjang asli.
    y = torch.tensor(frame['sentiment'].to_numpy(), dtype=torch.long)  # Tensor label kelas.
    return DataLoader(TensorDataset(X, lengths, y), batch_size=128, shuffle=shuffle)  # Membuat mini-batch.
train_loader = make_loader(train_df, shuffle=True)  # Batch training diacak.
val_loader = make_loader(val_df)  # Batch validasi tidak diacak.
test_loader = make_loader(test_df)  # Batch uji tidak diacak.
print('Ukuran train/validasi/test:', len(train_df), len(val_df), len(test_df))  # Untuk screenshot dan laporan.
print('Vocabulary:', len(vocab), '| MAX_LEN:', MAX_LEN)  # Rincian preprocessing.
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432025_AndriyanPaskaNanda
class SequenceModel(nn.Module):  # Definisi model LSTM untuk NLP.
    def __init__(self, vocab_size, embedding_dim=64, hidden_size=64, num_classes=3):  # Parameter arsitektur.
        super().__init__()  # Menginisialisasi nn.Module.
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)  # Indeks kata menjadi vektor.
        self.lstm = nn.LSTM(embedding_dim, hidden_size, batch_first=True)  # Memproses sekuens kata.
        self.dropout = nn.Dropout(0.3)  # Mengurangi risiko overfitting.
        self.fc = nn.Linear(hidden_size, num_classes)  # Memetakan hidden state ke tiga logit.
    def forward(self, x, lengths):  # Mendefinisikan forward propagation.
        embedded = self.embedding(x)  # Bentuk (batch, panjang, embedding_dim).
        packed = pack_padded_sequence(embedded, lengths.cpu(), batch_first=True, enforce_sorted=False)  # Abaikan padding.
        _, (hidden, _) = self.lstm(packed)  # Mengambil hidden state pada token terakhir yang valid.
        logits = self.fc(self.dropout(hidden[-1]))  # Regularisasi lalu proyeksi klasifikasi.
        return logits  # Logit mentah untuk CrossEntropyLoss.
```

  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432025_AndriyanPaskaNanda
EPOCHS, LEARNING_RATE = 6, 0.001  # Menetapkan jumlah epoch dan laju belajar.
model = SequenceModel(len(vocab)).to(DEVICE)  # Memindahkan model ke GPU/CPU.
print(model)  # Menampilkan arsitektur yang digunakan untuk dokumentasi laporan.
counts_train = np.bincount(train_df['sentiment'].to_numpy(), minlength=3)  # Menghitung kelas train saja.
class_weights = torch.tensor(len(train_df) / (3 * counts_train), dtype=torch.float32, device=DEVICE)  # Bobot inverse-frequency.
criterion = nn.CrossEntropyLoss(weight=class_weights)  # Mengatasi ketidakseimbangan kelas.
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)  # Optimizer untuk pembaruan bobot.
train_losses, val_losses, val_f1s = [], [], []  # Menyimpan riwayat loss dan macro-F1.
train_accs, val_accs = [], []  # Menyimpan riwayat akurasi tiap epoch.
best_f1, best_state = -1.0, None  # Menyimpan model terbaik menurut macro-F1 validasi.
for epoch in range(EPOCHS):  # Mengulang pelatihan beberapa epoch.
    model.train()  # Mengaktifkan mode training (termasuk dropout).
    total_loss, n_train = 0.0, 0  # Mengakumulasi loss rata-rata tertimbang sampel.
    train_hits = 0  # Menghitung prediksi benar dari data latih.
    for tokens, lengths, targets in train_loader:  # Mengambil mini-batch latih.
        tokens, targets = tokens.to(DEVICE), targets.to(DEVICE)  # Memindahkan tensor ke perangkat.
        optimizer.zero_grad()  # Menghapus gradien lama.
        logits = model(tokens, lengths)  # Forward propagation.
        loss = criterion(logits, targets)  # Menghitung loss klasifikasi.
        loss.backward()  # Backpropagation.
        optimizer.step()  # Memperbarui bobot LSTM.
        total_loss += loss.item() * len(targets); n_train += len(targets)  # Menjumlah loss terukur.
        train_hits += (logits.argmax(dim=1) == targets).sum().item()  # Menghitung jumlah label latih benar.
    train_losses.append(total_loss / n_train)  # Menyimpan rata-rata training loss.
    train_accs.append(train_hits / n_train)  # Menyimpan akurasi latih epoch ini.
    model.eval()  # Mematikan dropout saat validasi.
    total_val_loss, n_val, preds, truths = 0.0, 0, [], []  # Menyiapkan metrik validasi.
    with torch.no_grad():  # Tanpa gradien agar efisien.
        for tokens, lengths, targets in val_loader:  # Melintasi validasi saja.
            tokens, targets = tokens.to(DEVICE), targets.to(DEVICE)  # Memindahkan tensor.
            logits = model(tokens, lengths)  # Menghitung prediksi.
            total_val_loss += criterion(logits, targets).item() * len(targets); n_val += len(targets)  # Menambah loss.
            preds.extend(logits.argmax(dim=1).cpu().tolist()); truths.extend(targets.cpu().tolist())  # Mengumpulkan label.
    val_losses.append(total_val_loss / n_val)  # Mengarsipkan validation loss.
    f1 = f1_score(truths, preds, average='macro', zero_division=0)  # Menghitung macro-F1 validasi.
    val_f1s.append(f1)  # Menyimpan macro-F1 tiap epoch.
    val_accs.append(accuracy_score(truths, preds))  # Menyimpan akurasi validasi epoch ini.
    if f1 > best_f1:  # Memilih checkpoint terbaik tanpa test set.
        best_f1 = f1; best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}  # Salin bobot.
    print(f'Epoch {epoch+1:02d}/{EPOCHS} | Train loss={train_losses[-1]:.4f} | Val loss={val_losses[-1]:.4f} | Val acc={val_accs[-1]:.4f} | Val macro-F1={f1:.4f}')  # Log epoch.
model.load_state_dict(best_state)  # Mengembalikan bobot model terbaik.
model.eval()  # Menyiapkan evaluasi data uji yang belum pernah dilihat.
```

```python
#202432025_AndriyanPaskaNanda
plt.figure(figsize=(9, 4))  # Menyiapkan kanvas grafik loss.
plt.plot(range(1, EPOCHS + 1), train_losses, marker='o', label='Training loss', color='#114F6A')  # Menggambar loss latih.
plt.plot(range(1, EPOCHS + 1), val_losses, marker='s', label='Validation loss', color='#1FA4B1')  # Menggambar loss validasi.
plt.xlabel('Epoch'); plt.ylabel('CrossEntropy loss')  # Menuliskan label sumbu.
plt.title('Riwayat Training vs Validation Loss'); plt.legend(); plt.grid(alpha=0.3)  # Menyajikan informasi grafik.
plt.tight_layout(); plt.savefig('grafik_loss.png', dpi=160); plt.show()  # Menyimpan dan menampilkan grafik.
plt.figure(figsize=(9, 4))  # Menyiapkan grafik nilai macro-F1 validasi.
plt.plot(range(1, EPOCHS + 1), val_f1s, marker='o', color='#1FA4B1')  # Visualisasi macro-F1 per epoch.
plt.xlabel('Epoch'); plt.ylabel('Macro-F1'); plt.ylim(0, 1)  # Menetapkan skala metrik.
plt.title('Validasi Macro-F1'); plt.grid(alpha=0.3)  # Memberi konteks evaluasi.
plt.tight_layout(); plt.savefig('grafik_f1.png', dpi=160); plt.show()  # Menyimpan grafik F1.
plt.figure(figsize=(9, 4))  # Menyiapkan grafik akurasi setiap epoch.
plt.plot(range(1, EPOCHS + 1), train_accs, marker='o', label='Training accuracy', color='#114F6A')  # Akurasi latih.
plt.plot(range(1, EPOCHS + 1), val_accs, marker='s', label='Validation accuracy', color='#1FA4B1')  # Akurasi validasi.
plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.ylim(0, 1)  # Melabeli sumbu dan rentang akurasi.
plt.title('Riwayat Akurasi Training dan Validation'); plt.legend(); plt.grid(alpha=0.3)  # Memperjelas grafik.
plt.tight_layout(); plt.savefig('grafik_accuracy.png', dpi=160); plt.show()  # Menyimpan grafik akurasi.
test_true, test_pred, test_probs = [], [], []  # Menyiapkan hasil prediksi uji.
with torch.no_grad():  # Menghindari kalkulasi gradien saat testing.
    for tokens, lengths, targets in test_loader:  # Menelusuri data uji terpisah.
        logits = model(tokens.to(DEVICE), lengths)  # Menghitung output jaringan.
        probabilities = torch.softmax(logits, dim=1).cpu().numpy()  # Mengubah logit menjadi probabilitas.
        test_probs.extend(probabilities.tolist())  # Menyimpan probabilitas setiap kelas.
        test_pred.extend(probabilities.argmax(axis=1).tolist())  # Menyimpan label terprediksi.
        test_true.extend(targets.tolist())  # Menyimpan label aktual.
acc = accuracy_score(test_true, test_pred)  # Akurasi data uji.
macro_f1 = f1_score(test_true, test_pred, average='macro', zero_division=0)  # Macro-F1 data uji.
majority_baseline = max(np.bincount(test_true)) / len(test_true)  # Baseline kelas terbanyak.
print(f'TEST Accuracy={acc:.4f} | Macro-F1={macro_f1:.4f} | Baseline mayoritas={majority_baseline:.4f}')  # Ringkasan metrik.
print(classification_report(test_true, test_pred, target_names=LABELS, labels=[0,1,2], zero_division=0))  # Precision/recall/F1 per kelas.
cm = confusion_matrix(test_true, test_pred, labels=[0, 1, 2])  # Matriks benar/salah klasifikasi.
fig, ax = plt.subplots(figsize=(6, 5))  # Kanvas confusion matrix.
ConfusionMatrixDisplay(cm, display_labels=LABELS).plot(ax=ax, cmap='Blues', values_format='d', colorbar=False)  # Menampilkan matrix.
plt.title('Confusion Matrix Data Uji'); plt.tight_layout(); plt.savefig('confusion_matrix.png', dpi=160); plt.show()  # Menyimpan confusion matrix.
counts_pred = np.bincount(test_pred, minlength=3)  # Frekuensi kelas prediksi.
counts_actual = np.bincount(test_true, minlength=3)  # Frekuensi kelas aktual.
xpos = np.arange(3); width = 0.38  # Posisi grafik berdampingan.
plt.figure(figsize=(8, 4))  # Kanvas distribusi kelas.
plt.bar(xpos-width/2, counts_actual, width, label='Aktual', color='#114F6A')  # Batang jumlah aktual.
plt.bar(xpos+width/2, counts_pred, width, label='Prediksi', color='#1FA4B1')  # Batang jumlah prediksi.
plt.xticks(xpos, LABELS); plt.ylabel('Jumlah ulasan'); plt.title('Distribusi Aktual vs Prediksi')  # Label dan judul.
plt.legend(); plt.tight_layout(); plt.savefig('distribusi_prediksi.png', dpi=160); plt.show()  # Simpan grafik hasil.
results = pd.DataFrame({'ulasan': test_df['Review'].tolist(), 'aktual': [LABELS[v] for v in test_true], 'prediksi': [LABELS[v] for v in test_pred], 'skor_prediksi': np.max(np.array(test_probs), axis=1)})  # Menggabungkan hasil.
print(results[['aktual', 'prediksi', 'skor_prediksi']].head(10).to_string(index=False))  # Contoh skor keyakinan model.
results.to_csv('hasil_prediksi_uji.csv', index=False)  # Menyimpan hasil prediksi untuk analisis.

metrics = {'sumber': 'Trip Advisor Hotel Reviews (Kaggle)', 'total_data_setelah_cleaning': len(df), 'train_n': len(train_df), 'val_n': len(val_df), 'test_n': len(test_df), 'distribusi_sentimen': {LABELS[i]: int((df['sentiment']==i).sum()) for i in range(3)}, 'accuracy_test': float(acc), 'macro_f1_test': float(macro_f1), 'majority_baseline': float(majority_baseline), 'train_losses': train_losses, 'val_losses': val_losses, 'train_accuracy': train_accs, 'val_accuracy': val_accs, 'val_macro_f1': val_f1s, 'confusion_matrix': cm.tolist(), 'classification_report': classification_report(test_true, test_pred, target_names=LABELS, labels=[0,1,2], output_dict=True, zero_division=0)}  # Menghimpun seluruh metrik yang terukur.
with open('hasil_metrik.json', 'w', encoding='utf-8') as file: json.dump(metrics, file, indent=2, ensure_ascii=False)  # Menyimpan hasil riil untuk finalisasi laporan.
print('Metrik penelitian tersimpan: hasil_metrik.json')  # Mengonfirmasi hasil tersimpan.
# Mengemas grafik, metrik, dan prediksi menjadi satu ZIP agar mudah diunduh.
from zipfile import ZipFile  # Menggabungkan seluruh output praktikum.
output_files = ['grafik_loss.png', 'grafik_accuracy.png', 'grafik_f1.png', 'confusion_matrix.png', 'distribusi_prediksi.png', 'hasil_metrik.json', 'hasil_prediksi_uji.csv']  # File untuk pelaporan.
zip_name = 'HASIL_UTS_DEEPL_B_202432025.zip'  # Nama arsip keluaran.
with ZipFile(zip_name, 'w') as archive:  # Menyiapkan file ZIP hasil praktikum.
    for filename in output_files:  # Menambahkan seluruh file yang telah dihasilkan.
        if Path(filename).exists(): archive.write(filename)  # Memasukkan file grafik dan metrik.
print('\n=== SELESAI: hasil evaluasi asli tersedia ===')  # Menandai akhir eksekusi.
print('Silakan simpan screenshot output, unduh ZIP hasil, dan download notebook .ipynb.')  # Arah tindakan berikutnya.
from IPython.display import display, FileLink  # Membuat link unduh di output Colab.
display(FileLink(zip_name, result_html_prefix='UNDUH HASIL PRAKTIKUM: '))  # Alternatif tautan unduh di Jupyter/Colab.
try:  # Mengunduh arsip secara otomatis bila kode berjalan di Google Colab.
    from google.colab import files  # Fitur unduh browser Colab.
    files.download(zip_name)  # Memulai pengunduhan hasil tanpa langkah kode tambahan.
except ImportError:  # Bila notebook dijalankan di Jupyter lokal.
    print('ZIP hasil tersimpan pada folder notebook; unduh dari file yang tersedia.')  # Petunjuk pengguna Jupyter.
except Exception as exc:  # Bila browser mencegah unduhan otomatis.
    print('Unduhan otomatis tidak tersedia:', exc)  # Menjelaskan kendala tanpa menghentikan notebook.
    print('Klik ikon folder Files di kiri Colab, lalu unduh HASIL_UTS_DEEPL_B_202432025.zip.')  # Jalur alternatif.
```
